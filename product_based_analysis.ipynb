```python
import pandas as pd
import numpy as np
from itertools import combinations
from collections import Counter

# 1. Generate Transactional Data
np.random.seed(42)
products = ['Laptop', 'Wireless Mouse', 'Mechanical Keyboard', 'Monitor', 'Headphones', 'Desk Mat', 'Webcam']

data = []
# Simulate 500 unique orders
for order_id in range(1, 501):
    # Customers buy between 2 and 4 items per transaction
    num_items = np.random.randint(2, 5)
    # replace=False ensures a customer doesn't buy the exact same item twice in one order, automatically excluding self-pairs
    order_items = np.random.choice(products, size=num_items, replace=False)
    for item in order_items:
        data.append([order_id, item])

df_sales = pd.DataFrame(data, columns=['OrderID', 'Product'])

# 2. Market Basket Analysis
# Group products into a single list (basket) for each OrderID
baskets = df_sales.groupby('OrderID')['Product'].apply(list)

# Calculate pair combinations
pair_counts = Counter()
for basket in baskets:
    # Sort alphabetically so (Mouse, Keyboard) and (Keyboard, Mouse) count as the exact same pair
    basket = sorted(basket)
    pair_counts.update(combinations(basket, 2))

# 3. Output Top Pairs & Recommendations
df_top_pairs = pd.DataFrame(pair_counts.most_common(5), columns=['Product Pair', 'Frequency'])
print("--- TOP PRODUCT PAIRS ---")
print(df_top_pairs)
print("\n--- BUSINESS RECOMMENDATION ---")
print(f"Create a bundle deal for {df_top_pairs.iloc[0]['Product Pair'][0]} and {df_top_pairs.iloc[0]['Product Pair'][1]} to drive targeted cross-sell revenue.")
```
